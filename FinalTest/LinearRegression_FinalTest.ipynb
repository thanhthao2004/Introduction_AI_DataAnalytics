```python
df = pd.read_csv('boston.csv')
df.head()
```

```python
# Chọn các biến đầu vào và mục tiêu
X = df[['AGE', 'LSTAT']]
y = df['MEDV']
```

```python

# Chia dữ liệu thành tập huấn luyện và tập kiểm tra
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=10)

# Chuẩn hóa dữ liệu đầu vào
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Xây dựng mô hình hồi quy tuyến tính
model = LinearRegression()
model.fit(X_train_scaled, y_train)

# Dự đoán trên tập kiểm tra
y_pred = model.predict(X_test_scaled)

# Đánh giá mô hình
mae = mean_absolute_error(y_test, y_pred)
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print("Mean Absolute Error:", mae) # giá trị MAE độ chính xác của mô hình. giá trị thấp hơn cho thấy mô hình tốt hơn
print("Mean Squared Error:", mse) # giá trị MSE phạt nặng các sai số cho biết mức độ chính xác của mô hình. giá trị thấp hơn cho thấy mô hình tốt hơn
print("R-squared:", r2) # giá trị R2 cho biết mức độ mà mô hình giải thích phân trăm biến thiên. giá trị càng gần 1 cho thấy mô hình tốt hơn
```

```python
X_test=np.arange(0,len(X_test),1)
plt.scatter(X_test, y_test,  color='blue')
plt.plot(X_test, y_pred, color='red', linewidth=3)
plt.show()
```
