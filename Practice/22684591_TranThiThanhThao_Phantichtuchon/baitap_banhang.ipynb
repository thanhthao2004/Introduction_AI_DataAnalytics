```python
# 33. Hàm phân loại
def foo(x):
    if x < 10:
        return 'BAD'
    elif 10 <= x < 50:
        return 'GOOD'
    else:
        return 'EXCELLENT'
df['WORTH'] = df[['QUANTITYORDERED']].applymap(foo)
df[['QUANTITYORDERED','WORTH']]
```

```python
# 34. Ánh xạ giá trị tới cột QTR_ID
map_dict = {1:'Qui_1',2:'Qui_2',3:'Qui_3',4:'Qui_4'}
df['QTR_ID'] = df['QTR_ID'].map(map_dict)
df['QTR_ID']
```
