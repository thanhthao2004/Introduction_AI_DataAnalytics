```python
# Import các thư viện cần thiết
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import statistics
from scipy.stats import norm, binom, poisson
import seaborn as sns

# Thiết lập style cho matplotlib
plt.style.use('seaborn-v0_8')
plt.rcParams['figure.figsize'] = (12, 8)
plt.rcParams['font.size'] = 12

print("Đã import thành công các thư viện cần thiết!")
```

## 1. Thống kê mô tả (Central Tendency)

### Dữ liệu mẫu:
Chúng ta sẽ sử dụng dữ liệu: [19, 15, 20, 25, 35, 39, 49, 45, 50, 55]


```python
# Tạo dữ liệu mẫu
data = np.array([19, 15, 20, 25, 35, 39, 49, 45, 50, 55])
print(f"Dữ liệu: {data}")
print(f"Kích thước mẫu: {len(data)}")
```

### 1.1. Mean (Trung bình)


```python
# Tính trung bình
mean = np.mean(data)
print(f"Trung bình của dữ liệu: {mean:.2f}")
```

### 1.2. Median (Trung vị)


```python
# Tính trung vị
sorted_data = np.sort(data)
median = np.median(sorted_data)
print(f"Dữ liệu đã sắp xếp: {sorted_data}")
print(f"Trung vị của dữ liệu: {median:.2f}")
```

### 1.3. Mode (Yếu vị)


```python
# Tính yếu vị (mode)
mode_value = statistics.mode(data)
print(f"Yếu vị của dữ liệu: {mode_value}")
```

### 1.4. Percentile (Phân vị)


```python
# Tính phân vị thứ 25 (Q1)
p = 25
percentile_value = np.percentile(data, p)
print(f"Phân vị thứ {p} của dữ liệu: {percentile_value}")
```

### 1.5. Quartile (Tứ phân vị)


```python
# Tính các tứ phân vị
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)  # Trung vị
q3 = np.percentile(data, 75)

print(f"Tứ phân vị Q1: {q1}")
print(f"Tứ phân vị Q2 (Trung vị): {q2}")
print(f"Tứ phân vị Q3: {q3}")
```

## 2. Độ đo phân tán (Variability)


### 2.1. Khoảng biến thiên (Range)


```python
# Tính khoảng biến thiên
range_value = np.max(data) - np.min(data)
print(f"Khoảng biến thiên: {range_value}")
```

### 2.2. Khoảng tứ phân vị (Interquartile Range - IQR)


```python
# Tính IQR
iqr = q3 - q1
print(f"Khoảng tứ phân vị (IQR): {iqr}")
```

### 2.3. Độ lệch chuẩn (Standard Deviation)


```python
# Tính độ lệch chuẩn
std_deviation = np.std(data)
print(f"Độ lệch chuẩn của dữ liệu: {std_deviation}")
```

## 3. Phân phối chuẩn (Normal Distribution)


### 3.1. Hàm mật độ xác suất (PDF) của phân phối chuẩn


```python
# Vẽ hàm mật độ của phân phối chuẩn với các tham số khác nhau
x = np.linspace(150, 190, 100)
y1 = norm.pdf(x, loc=170, scale=3)  # μ=170, σ=3
y2 = norm.pdf(x, loc=170, scale=5)  # μ=170, σ=5
y3 = norm.pdf(x, loc=160, scale=3)  # μ=160, σ=3

plt.figure(figsize=(12, 6))
plt.plot(x, y1, label='$\\mu=170, \\sigma=3$')
plt.plot(x, y2, label='$\\mu=170, \\sigma=5$')
plt.plot(x, y3, label='$\\mu=160, \\sigma=3$')
plt.title('HÀM MẬT ĐỘ PHÂN PHỐI CHUẨN')
plt.xlabel('Giá trị')
plt.ylabel('Mật độ xác suất')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

### 3.2. Phân phối chuẩn chuẩn hóa (μ=0, σ=1)


```python
# Vẽ phân phối chuẩn chuẩn hóa
xx = np.linspace(-3, 3, 100)
yy = norm.pdf(xx)  # Mặc định loc=0, scale=1

plt.figure(figsize=(10, 6))
plt.plot(xx, yy, color='darkorange', linewidth=2, label='$\\mu=0, \\sigma=1$')
plt.title('HÀM MẬT ĐỘ PHÂN PHỐI CHUẨN CHUẨN HÓA')
plt.xlabel('Giá trị')
plt.ylabel('Mật độ xác suất')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()

# Tính giá trị PDF tại x=1
pdf_at_1 = norm.pdf(1)
print(f"Giá trị PDF tại x=1: {pdf_at_1}")
```

### 3.3. Quy tắc 68-95-99.7 trên phân phối chuẩn


```python
# Tính xác suất theo quy tắc 68-95-99.7
xs_be_hon_1 = norm.cdf(x=-1, loc=0, scale=1)  # P(X < -1)
xs_lon_hon_1 = 1 - norm.cdf(x=1, loc=0, scale=1)  # P(X > 1)
xs_giua = 1 - (xs_be_hon_1 + xs_lon_hon_1)  # P(-1 < X < 1)

print(f"P(X < -1): {xs_be_hon_1}")
print(f"P(X > 1): {xs_lon_hon_1}")
print(f"P(-1 < X < 1): {xs_giua}")

# Vẽ đồ thị minh họa quy tắc 68-95-99.7
x_range = np.arange(-4, 4, 0.01)
y_range = norm.pdf(x_range)

plt.figure(figsize=(12, 6))
plt.plot(x_range, y_range, 'b-', linewidth=2, label='Phân phối chuẩn')

# Tô màu các vùng
plt.fill_between(x=np.arange(-4, -1, 0.01), 
                 y1=norm.pdf(np.arange(-4, -1, 0.01)), 
                 color='red', alpha=0.35, label='P(X < -1)')
plt.fill_between(x=np.arange(1, 4, 0.01), 
                 y1=norm.pdf(np.arange(1, 4, 0.01)), 
                 color='red', alpha=0.35, label='P(X > 1)')
plt.fill_between(x=np.arange(-1, 1, 0.01), 
                 y1=norm.pdf(np.arange(-1, 1, 0.01)), 
                 color='blue', alpha=0.35, label='P(-1 < X < 1)')

# Thêm text hiển thị xác suất
plt.text(x=-1.9, y=0.03, s=round(xs_be_hon_1, 3), fontsize=12, fontweight='bold')
plt.text(x=-0.4, y=0.1, s=round(xs_giua, 3), fontsize=12, fontweight='bold')
plt.text(x=1.3, y=0.03, s=round(xs_lon_hon_1, 3), fontsize=12, fontweight='bold')

plt.title('QUY TẮC 68-95-99.7 TRÊN PHÂN PHỐI CHUẨN')
plt.xlabel('Giá trị')
plt.ylabel('Mật độ xác suất')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## 4. Các hàm thống kê quan trọng


### 4.1. Hàm CDF (Cumulative Distribution Function)


```python
# Tính CDF
cdf_minus_2 = norm.cdf(-2).round(5)
cdf_8_normal = norm.cdf(8, loc=16, scale=4).round(5)

print(f"CDF(-2) với phân phối chuẩn chuẩn: {cdf_minus_2}")
print(f"CDF(8) với μ=16, σ=4: {cdf_8_normal}")
```

### 4.2. Hàm PPF (Percent Point Function) - Hàm ngược của CDF


```python
# Tính PPF
ppf_092275 = norm.ppf(0.92275, loc=16, scale=4).round(2)
ppf_standard = norm.ppf(0.92275).round(2)

print(f"PPF(0.92275) với μ=16, σ=4: {ppf_092275}")
print(f"PPF(0.92275) với phân phối chuẩn chuẩn: {ppf_standard}")
```

### 4.3. Hàm SF (Survival Function) - Hàm sống sót


```python
# Tính SF
sf_2 = norm.sf(2).round(5)
print(f"SF(2) = P(X > 2): {sf_2}")
```

### 4.4. Hàm ISF (Inverse Survival Function)


```python
# Tính ISF
isf_092275 = norm.isf(0.92275).round(2)
print(f"ISF(0.92275): {isf_092275}")
```

### 4.5. Ví dụ ứng dụng thực tế


```python
# Ví dụ: Chiều cao của nam giới có phân phối chuẩn với μ=175cm, σ=4cm
# Tính xác suất một người nam có chiều cao từ 166cm đến 177cm

mu = 175  # Chiều cao trung bình
sigma = 4  # Độ lệch chuẩn

# Tính xác suất
x177 = norm.cdf(177, loc=mu, scale=sigma)  # P(X ≤ 177)
x166 = norm.cdf(166, loc=mu, scale=sigma)  # P(X ≤ 166)
prob_between = x177 - x166  # P(166 ≤ X ≤ 177)

print(f"Xác suất chiều cao từ 166cm đến 177cm: {prob_between:.4f}")
print(f"Tức là {prob_between*100:.2f}% nam giới có chiều cao trong khoảng này")
```

### 4.6. Tổng kết các thống kê mô tả


```python
# Tạo bảng tổng kết
summary_stats = {
    'Thống kê': ['Mean', 'Median', 'Mode', 'Q1', 'Q2', 'Q3', 'Range', 'IQR', 'Std Dev'],
    'Giá trị': [mean, median, mode_value, q1, q2, q3, range_value, iqr, std_deviation]
}

df_summary = pd.DataFrame(summary_stats)
print("=== TỔNG KẾT THỐNG KÊ MÔ TẢ ===")
print(df_summary.to_string(index=False))
```
