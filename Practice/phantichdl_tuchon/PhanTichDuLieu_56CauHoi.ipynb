```python
# Import các thư viện cần thiết
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import random
from datetime import datetime

# Thiết lập hiển thị
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)
plt.rcParams['figure.figsize'] = (12, 8)

print("Đã import thành công các thư viện cần thiết!")
```

## CÂU HỎI 1-10: ĐỌC VÀ HIỂN THỊ DỮ LIỆU CƠ BẢN


```python
# CÂU HỎI 1: Đọc File với các file trong thư mục dữ liệu. Hiển thị toàn bộ dữ liệu của file dữ liệu đã đọc
print("=== CÂU HỎI 1: ĐỌC VÀ HIỂN THỊ TOÀN BỘ DỮ LIỆU ===")
df = pd.read_csv('Social Media Data.csv')
print("Đã đọc thành công file Social Media Data.csv")
print(f"Kích thước dữ liệu: {df.shape}")
print("\nToàn bộ dữ liệu:")
print(df)
print("\n*** GIẢI THÍCH: Dữ liệu chứa thông tin về các công ty với các chỉ số mạng xã hội, doanh thu, vốn hóa thị trường và thông tin khác. Có {0} dòng và {1} cột dữ liệu.***".format(df.shape[0], df.shape[1]))
```

```python
# CÂU HỎI 2: Đọc file với chỉ định không có header, dòng header của chúng ta đã biến thành 1 bản ghi dữ liệu
print("=== CÂU HỎI 2: ĐỌC FILE KHÔNG CÓ HEADER ===")
df_no_header = pd.read_csv('Social Media Data.csv', header=None)
print("Đã đọc file không có header")
print(f"Kích thước dữ liệu: {df_no_header.shape}")
print("\n5 dòng đầu tiên:")
print(df_no_header.head())
print("\n*** GIẢI THÍCH: Khi đọc không có header, dòng đầu tiên (tên cột) trở thành dữ liệu thực tế. Pandas tự động gán tên cột là số từ 0 đến n.***")
```

```python
# CÂU HỎI 3: Đọc File với các tùy chọn mặc định. Hiển thị 5 dòng dữ liệu đầu tiên
print("=== CÂU HỎI 3: HIỂN THỊ 5 DÒNG ĐẦU TIÊN ===")
print("5 dòng dữ liệu đầu tiên:")
print(df.head())
print("\n*** GIẢI THÍCH: head() hiển thị 5 dòng đầu tiên của DataFrame, giúp ta có cái nhìn tổng quan về cấu trúc và nội dung dữ liệu.***")
```

```python
# CÂU HỎI 4: Đọc File với các tùy chọn mặc định. Hiển thị 5 dòng dữ liệu cuối cùng
print("=== CÂU HỎI 4: HIỂN THỊ 5 DÒNG CUỐI CÙNG ===")
print("5 dòng dữ liệu cuối cùng:")
print(df.tail())
print("\n*** GIẢI THÍCH: tail() hiển thị 5 dòng cuối cùng của DataFrame, giúp ta kiểm tra dữ liệu ở cuối file và đảm bảo dữ liệu được đọc đầy đủ.***")
```

```python
# CÂU HỎI 5: Chuyển kiểu dữ liệu cho 1 cột nào đó
print("=== CÂU HỎI 5: CHUYỂN KIỂU DỮ LIỆU ===")
print("Kiểu dữ liệu ban đầu của cột 'Follower Count':")
print(df['Follower Count'].dtype)
print("\nChuyển đổi 'Follower Count' từ object sang numeric:")
df['Follower Count'] = df['Follower Count'].str.replace(',', '').astype(int)
print("Kiểu dữ liệu sau khi chuyển đổi:")
print(df['Follower Count'].dtype)
print("\n5 giá trị đầu tiên sau chuyển đổi:")
print(df['Follower Count'].head())
print("\n*** GIẢI THÍCH: Chuyển đổi kiểu dữ liệu giúp thực hiện các phép tính toán học trên cột. Từ string sang int để có thể tính toán số học.***")
```

```python
# CÂU HỎI 6: Xem chiều dài của df, tương đương shape[0]
print("=== CÂU HỎI 6: CHIỀU DÀI CỦA DATAFRAME ===")
print(f"Chiều dài của df: {len(df)}")
print(f"Shape[0]: {df.shape[0]}")
print(f"Kết quả so sánh: {len(df) == df.shape[0]}")
print("\n*** GIẢI THÍCH: len(df) và df.shape[0] đều trả về số dòng trong DataFrame. len() là hàm Python built-in, shape[0] là thuộc tính của pandas.***")
```

```python
# CÂU HỎI 7: Xem thông tin dataframe vừa đọc được
print("=== CÂU HỎI 7: THÔNG TIN DATAFRAME ===")
print("Thông tin chi tiết về DataFrame:")
df.info()
print("\n*** GIẢI THÍCH: df.info() cung cấp thông tin tổng quan về DataFrame bao gồm: số dòng, số cột, tên cột, kiểu dữ liệu, và số giá trị non-null. Giúp hiểu cấu trúc dữ liệu.***")
```

```python
# CÂU HỎI 8: Xem kích thước của dataframe
print("=== CÂU HỎI 8: KÍCH THƯỚC DATAFRAME ===")
print(f"Shape: {df.shape}")
print(f"Số dòng: {df.shape[0]}")
print(f"Số cột: {df.shape[1]}")
print("\n*** GIẢI THÍCH: df.shape trả về tuple (số_dòng, số_cột). Giúp hiểu quy mô của dataset.***")
```

```python
# CÂU HỎI 9: Hiển thị dữ liệu của cột thứ 3
print("=== CÂU HỎI 9: DỮ LIỆU CỘT THỨ 3 ===")
print("Tên các cột:")
print(df.columns.tolist())
print(f"\nCột thứ 3: '{df.columns[2]}'")
print("Dữ liệu của cột thứ 3:")
print(df.iloc[:, 2])
print("\n*** GIẢI THÍCH: iloc[:, 2] lấy tất cả dòng của cột thứ 3 (index 2). Cột thứ 3 là 'Social Media Presence'.***")
```

```python
# CÂU HỎI 10: Hiển thị dữ liệu của cột 1,2,3,5,6
print("=== CÂU HỎI 10: DỮ LIỆU NHIỀU CỘT ===")
print("Tên các cột được chọn:")
selected_columns = [df.columns[0], df.columns[1], df.columns[2], df.columns[4], df.columns[5]]
print(selected_columns)
print("\nDữ liệu của các cột được chọn:")
print(df[selected_columns])
print("\n*** GIẢI THÍCH: Chọn nhiều cột bằng cách truyền list tên cột vào df[]. Giúp xem dữ liệu của các cột quan trọng cùng lúc.***")
```
